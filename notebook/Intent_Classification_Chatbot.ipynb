### Preprocessing (Tokenization)

Tokenization is the process of converting raw text into smaller units called tokens. as computers cannot understand raw text directly we must then convert: `Text → Tokens → Numbers → Model`

There are three type:
1. Word tokenization: split by spaces
    - TF-IDF,
    - Bag of Words,
    - Logistic Regression etc...
2. Subword Tokenization: Break words into smaller meaningful parts (it's what model like BERT use)
    - e.g unbelievable ---> "un", "believ", "able"
3. Character Tokenization:
    - "book" → ["b", "o", "o", "k"] (not effective and it is rarely used)

</br>

**How Tokenization Works Internally**
- Step 1: Normalize text
  ```
  "I LOVE NLP!!!"
→ "i love nlp!!!"
  ```
- Step 2: Split into subwords
  ```
  "playing" → ["play", "##ing"]
  ```
- Step 3: Convert to IDs
  ```
  ["play", "##ing"] → [2345, 6789]
  ```
- Step 4: Add special tokens
  ```
  [CLS] I want to book [SEP]
  ```

</br>

Tokenization is NOT just splitting text. **It is how the model understands language**

## Building a Traditional ML model baseline

**Logistic Regression Model**
because Logistic Regression is: the strongest classical NLP baseline, simple mathematically and it is surprisingly powerful for text classification

````python
import re

from sklearn.feature_extraction.text import TfidfVectorizer

import torch
````

````python
# extracting Text and Labels
train_texts = dataset["train"]["text"]
train_labels = dataset["train"]["intent"]

val_texts = dataset["validation"]["text"]
val_labels = dataset["validation"]["intent"]

test_texts = dataset["test"]["text"]
test_labels = dataset["test"]["intent"]
````

````python
# preprocessing
def preprocess_for_tfidf(text):
    cleaned_text = text.lower()
    cleaned_text = re.sub(r"[^\w\s]", "", text)  # removing punctuation
    return cleaned_text
````

````python
train_texts
````

````python
train_texts = [preprocess_for_tfidf(t) for t in train_texts]
val_texts = [preprocess_for_tfidf(t) for t in val_texts]
test_texts = [preprocess_for_tfidf(t) for t in test_texts]
````

````python
train_texts
````

````python
# TF-IDF Vectorization
vectorizer = TfidfVectorizer(
    max_features = 5000, # limiting vocabulary size
    ngram_range = (1, 2) # considering unigrams and bigrams together
)
````

````python
vectorizer
````

````python
# fitting vectorizer to our dataset in other to learn it token representation value vector :
# learning vocabulary and importance of words

# we fit on training dataset only
X_train = vectorizer.fit_transform(train_texts)
````

````python
# transforming validation and testing dataset
X_val = vectorizer.transform(val_texts)
X_test = vectorizer.transform(test_texts)
````

````python
type(X_train) # we might need to sparse X_train later as sparse matrix most values are 0
````

````python
print(X_train.shape)
````

````python
# converting to PyTorch tensors

X_train_tensor = torch.tensor(X_train.toarray(), dtype=torch.float32)
X_val_tensor = torch.tensor(X_val.toarray(), dtype=torch.float32)
X_test_tensor = torch.tensor(X_test.toarray(), dtype=torch.float32)

y_train_tensor = torch.tensor(train_labels, dtype=torch.long)
y_val_tensor = torch.tensor(val_labels, dtype=torch.long)
y_test_tensor = torch.tensor(test_labels, dtype=torch.long)
````

`toarray()` converts sparse to **dense**. for this dataset it is ok because the dataset is manageable but in large datasets it can cause memory issue

````python
# creating PyTorch Dataset
from torch.utils.data import Dataset, DataLoader

class IntentDataset(Dataset):
  def __init__(self, X, y):
    self.X = X
    self.y = y

  def __len__(self):
    return len(self.y)

  def __getitem__(self, idx):
    return self.X[idx], self.y[idx]
````

````python
train_dataset = IntentDataset(X_train_tensor, y_train_tensor)
val_dataset = IntentDataset(X_val_tensor, y_val_tensor)
````

````python
len(train_dataset), len(val_dataset)
````

````python
train_dataset[0], val_dataset[0]
````

````python
# dataloading, applying DataLoader
train_loader = DataLoader(
                  train_dataset,
                  batch_size = 32,
                  shuffle = True
               )
val_loader = DataLoader(
                  val_dataset,
                  batch_size = 32
             )
````

````python
# building the Logistic Regression Model
import torch
import torch.nn as nn
````

````python
## defining the model
class LogisticRegressionModel(nn.Module):
    '''
      input_dim: number of TF-IDF features
      num_classes: Number of intent labels

      applying:
        nn.Linear: This is the actual Logistic Regression equation: y = Wx + b
                  x: TF-IDF vector
                  W: learned weights
                  b: bias
    '''
    def __init__(self, input_dim, num_classes):
        super().__init__()

        self.linear = nn.Linear(input_dim, num_classes)

    def forward(self, x):
        outputs = self.linear(x)
        return outputs
````

````python
## initializing the model
input_dim = X_train_tensor.shape[1]
num_classes = len(class_names)

model = LogisticRegressionModel(input_dim, num_classes)
````

````python
model
````

````python
## defining criterion or loss function and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = 0.001)
````

````python
## training loop

num_epochs = 30

for epoch in range(num_epochs):
    model.train()
    total_loss = 0

    for inputs, labels in train_loader:
        optimizer.zero_grad()

        outputs = model(inputs)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)

    print(f"Epoch [{epoch+1}/{num_epochs}] Loss: {avg_loss:.4f}")
````

````python
# evaluation

model.eval()

correct = 0
total = 0

with torch.no_grad():

    for inputs, labels in val_loader:
        outputs = model(inputs)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)

        correct += (predicted == labels).sum().item()
````

````python
# final accuracy

accuracy = 100 * correct / total
print(f"Validation Accuracy: {accuracy:.2f}%")
````

````python
# Evaluation analysis

## Evaluation Prediction Loop
all_preds = []
all_labels = []

model.eval()

with torch.no_grad():

    for inputs, labels in val_loader:
        outputs = model(inputs)

        _, predicted = torch.max(outputs, 1)

        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
````

````python
import re

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
````

````python
# extracting Text and Labels

train_texts = dataset["train"]["text"]
train_labels = dataset["train"]["intent"]

val_texts = dataset["validation"]["text"]
val_labels = dataset["validation"]["intent"]

test_texts = dataset["test"]["text"]
test_labels = dataset["test"]["intent"]
````

````python
# preprocessing

def preprocess_for_tfidf(text):
    cleaned_text = text.lower()
    cleaned_text = re.sub(r"[^\w\s]", "", text)  # removing punctuation
    return cleaned_text
````

````python
# TF-IDF Vectorization

vectorizer = TfidfVectorizer(
    max_features = 5000,
    ngram_range = (1, 2)
)
````

````python
# fitting vectorizer to our dataset

X_train = vectorizer.fit_transform(train_texts)

# transforming validation and testing dataset

X_val = vectorizer.transform(val_texts)
X_test = vectorizer.transform(test_texts)
````

````python
# converting to PyTorch tensors

X_train_tensor = torch.tensor(X_train.toarray(), dtype=torch.float32)
X_val_tensor = torch.tensor(X_val.toarray(), dtype=torch.float32)
X_test_tensor = torch.tensor(X_test.toarray(), dtype=torch.float32)

y_train_tensor = torch.tensor(train_labels, dtype=torch.long)
y_val_tensor = torch.tensor(val_labels, dtype=torch.long)
y_test_tensor = torch.tensor(test_labels, dtype=torch.long)
````

````python
# creating PyTorch Dataset

class IntentDataset(Dataset):
  def __init__(self, X, y):
    self.X = X
    self.y = y

  def __len__(self):
    return len(self.y)

  def __getitem__(self, idx):
    return self.X[idx], self.y[idx]
````

````python
train_dataset = IntentDataset(X_train_tensor, y_train_tensor)
val_dataset = IntentDataset(X_val_tensor, y_val_tensor)
````

````python
# dataloading, applying DataLoader

train_loader = DataLoader(
                  train_dataset,
                  batch_size = 32,
                  shuffle = True
               )
val_loader = DataLoader(
                  val_dataset,
                  batch_size = 32
             )
````

````python
# building the model

class IntentMLP(nn.Module):
    '''
    512 hidden units taken
    nn.ReLU() activation function for adding non-linearity
    '''
    def __init__(self, input_dim, num_classes):

        super().__init__()

        self.linear1 = nn.Linear(input_dim, 512)
        self.relu1 = nn.ReLU()

        self.linear2 = nn.Linear(512, 256)
        self.relu2 = nn.ReLU()

        self.linear3 = nn.Linear(256, num_classes)

    def forward(self, x):
        x = self.linear1(x)
        print(f"Linear 1 output shape: {x.shape}")
        x = self.relu1(x)
        print(f"ReLU 1 output shape: {x.shape}")
        x = self.linear2(x)
        print(f"Linear 2 output shape: {x.shape}")
        x = self.relu2(x)
        print(f"ReLU 2 output shape: {x.shape}")
        x = self.linear3(x)
        return x
````

````python
class IntentMLP(nn.Module):
    '''
    512 hidden units taken
    '''
    def __init__(self, input_dim, num_classes):

        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),

            nn.Linear(512, 256),
            nn.ReLU(),

            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        return self.network(x)
````

````python
# initializing the model

input_dim = X_train_tensor.shape[1]
num_classes = len(class_names)

model = IntentMLP(
    input_dim = input_dim,
    num_classes = num_classes
)
````

````python
# defining loss function and optimizer

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = 0.001)
````

````python
train_losses = []
val_accuracies = []
````

````python
# validation function

def evaluate(model, dataloader):

    model.eval()

    correct = 0
    total = 0

    with torch.no_grad():

        for inputs, labels in dataloader:

            outputs = model(inputs)

            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)

            correct += (predicted == labels).sum().item()

    accuracy = 100 * correct / total

    return accuracy
````

````python
# training loop

num_epochs = 30

for epoch in range(num_epochs):

    model.train()

    total_loss = 0

    for inputs, labels in train_loader:

        optimizer.zero_grad()

        outputs = model(inputs)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)

    train_losses.append(avg_loss)

    val_acc = evaluate(model, val_loader)

    val_accuracies.append(val_acc)

    print(
        f"Epoch {epoch+1}/{num_epochs} | "
        f"Loss: {avg_loss:.4f} | "
        f"Val Acc: {val_acc:.2f}%"
    )
````

````python
# validation accuracy

plt.figure(figsize=(8,5))

plt.plot(
    range(1, len(val_accuracies)+1),
    val_accuracies,
    marker = 'o'
)

plt.title("Validation Accuracy Curve")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")

plt.grid(True)

plt.show()
````

```
meanwhile the MLP neural network model is learning unnecessary complexity. TF-IDF has already done much of the feature engineering.

## BERT Fine-Tuning: DistilBERT

**DistilBERT** instead of full **BERT** to leverage less memory and easier training while achieving similar performance on intent classification

</br>

BERT stands for Bidirectional Encoder Representations from Transformers. It was created to replace architecture based on RNN, LTSM, GRU

**why transformers were revolutionary?**
suppose we have:
```
I booked a flight to Delhi.
```
RNN reads:
```
I → booked → a → flight → to → Delhi
```
one word at a time.

A Transformer processes:
```
I
booked
a
flight
to
Delhi
```
all at once. This allows it to learn relationships between any words in the sentence efficiently.

### TF-IDF vs BERT

**1. TF-IDF**

Sentence: ```I want to book a flight```

TF-IDF might produce:
| Word   | Score |
| ------ | ----- |
| book   | 0.72  |
| flight | 0.83  |
| want   | 0.22  |

The model sees:
```
book
flight
want
```
mostly as independent features.

Limitation

Consider: ```I want to book a flight```

and ```I want to read a book```

TF-IDF sees: `book` in both.

It has no understanding that: `book = reservation` in one sentence and `book = object` to read in another.

</br>

**2. BERT**

BERT creates a representation that depends on context.

The word: `book` gets different representations depending on surrounding words.

Example: `book a flight` and `read a book`

The vector for `book` will be different. This is one of the major breakthroughs of Transformers.

### Subword Tokenization

Example: `unbelievable`

BERT tokenizer might split into: `["un", "##believ", "##able"]` or something similar.

</br>

             Text

              ↓
        Classifier Head
              ↓

```

### Concept

Right now your TF-IDF vector is something like: `[0.0, 0.3, 0.7, 0.0, ...]` A fixed feature vector.

With BERT, every word gets a learned contextual representation. Conceptually: `book` might become `[0.23, -0.44, 1.12, ...]` a vector with hundreds of dimensions. And that vector changes depending on context.

**Important concept before Training**:

With the MLP, we trained for: `30 epochs`

With DistilBERT, we typically do: `2-5 epochs` not `30`.

</br>

Because we're not learning language from scratch. We're only slightly adjusting a pretrained model. Training too long can cause: **catastrophic forgetting**, where the model starts losing some of the useful knowledge acquired during pretraining.

````python
# defining hyperparameters

batch_size = 16
learning_rate = 2e-5
epochs = 3
````

````python
# validation function

def evaluate(model, dataloader, device):

    model.eval()

    correct = 0
    total = 0

    with torch.no_grad():

        for batch in dataloader:
            batch = {k: v.to(device) for k, v in batch.items() if k != "token_type_ids"}

            outputs = model(
                input_ids = batch["input_ids"],
                attention_mask = batch["attention_mask"]
            )

            predictions = outputs.logits.argmax(dim = 1)

            correct += (predictions == batch["labels"]).sum().item()

            total += batch["labels"].size(0)

    return 100 * correct / total
````

````python
# defining tracking lists

train_losses = []
val_accuracies = []
````

````python
# training loop

num_epochs = 3

for epoch in range(num_epochs):

    model.train()

    total_loss = 0

    for batch in train_loader:
        batch = {k: v.to(device) for k, v in batch.items() if k != "token_type_ids"}

        optimizer.zero_grad()

        outputs = model(
            input_ids = batch["input_ids"],
            attention_mask = batch["attention_mask"],
            labels = batch["labels"]
        )

        loss = outputs.loss

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)

    train_losses.append(avg_loss)

    val_acc = evaluate(
        model,
        val_loader,
        device
    )

    val_accuracies.append(val_acc)

    print(
        f"Epoch {epoch+1}/{num_epochs} | "
        f"Loss: {avg_loss:.4f} | "
        f"Val Acc: {val_acc:.2f}%"
    )
````

### Error Analysis

we should learn why mistakes happen

````python
misclassified = []

model.eval()

# i removed earlier text column from the tokenized_dataset so that explain the workaround

# Assuming dataset["test"]["text"] still holds the original text in order
# and test_loader is not shuffled (default for DataLoader if not specified)
original_test_texts = dataset["test"]["text"]
current_idx = 0

with torch.no_grad():

    for batch in test_loader:

        batch_labels = batch["labels"].cpu().numpy()
        batch_size = len(batch_labels)

        batch_gpu = {
            k: v.to(device)
            for k, v in batch.items()
            if k not in ["text", "token_type_ids"] # "text" is already removed from batch
        }

        outputs = model(
            input_ids = batch_gpu["input_ids"],
            attention_mask = batch_gpu["attention_mask"]
        )

        predictions = (
            outputs.logits
            .argmax(dim=1)
            .cpu()
            .numpy()
        )

        for i in range(batch_size):
            text = original_test_texts[current_idx + i] # Retrieve original text
            true_label = batch_labels[i]
            pred_label = predictions[i]

            if true_label != pred_label:

                misclassified.append({
                    "text": text,
                    "true": label_names[true_label],
                    "pred": label_names[pred_label]
                })
        current_idx += batch_size
````

**Confusion Matrix Analysis**

Analysis of the confusion matrix revealed that the majority of misclassifications involved the Out-of-Scope (OOS) category. The most common errors occurred when OOS utterances were incorrectly classified as domain-specific intents such as "smart_home", "weather", "schedule_maintenance", "order", and "jump_start".

For example, OOS samples were misclassified as "smart_home" 22 times and as "weather" 18 times. This behavior suggests that the model attempts to associate unknown user requests with the most semantically similar intent rather than confidently assigning the OOS label.

Among the in-scope intents, the most frequent confusion was observed between "redeem_rewards" and "rewards_balance". These intents belong to the same financial domain and contain highly similar vocabulary, making them difficult to distinguish. This indicates that while DistilBERT captures domain semantics effectively, it may struggle when two intents share nearly identical linguistic patterns.

The confusion matrix further demonstrates that most errors occur between semantically related intents rather than unrelated categories. This observation is encouraging because it indicates that the model has learned meaningful intent representations and generally understands the underlying domains of user requests.

Future improvements could focus on strengthening OOS detection and providing additional training examples for highly similar intent pairs to improve class separation.

</br>

**Error Analysis**

Manual inspection of misclassified examples provided further insight into the model's limitations.

Several errors were caused by semantic overlap between related intents. For example, the utterance "how much money can i transfer today" was incorrectly classified as "income" instead of "transfer". Both intents belong to the financial domain and involve monetary transactions, which likely contributed to the confusion.

Similarly, multiple "payday" examples such as "what day will i earn my payment from work" and "when will my payment be deposited" were predicted as "bill_due". In these cases, the model recognized the financial context but failed to distinguish between incoming payments and outgoing obligations.

The model also struggled with specialized banking intents. Several utterances related to "replacement_card_duration" were classified as "new_card" or "make_call". These predictions indicate that the model correctly identified the credit-card domain but misunderstood the specific user objective.

A second category of errors involved OOS predictions. Questions such as "what is the definition of auspicious" and "is there a reason beyond biology about why humans are alive" were classified as OOS rather than their respective intents. These examples suggest that the model sometimes lacks confidence when encountering abstract, philosophical, or uncommon formulations.

Additional confusion was observed between operationally related intents. For example, "vacation request please" was classified as "pto_request_status" rather than "pto_request", indicating difficulty distinguishing between creating a request and checking its status.

Overall, the error analysis reveals that the model rarely confuses completely unrelated domains. Instead, most errors arise from semantically similar intents, ambiguous wording, and challenges associated with OOS detection. These findings suggest that future improvements should focus on hard-negative training examples, improved OOS calibration, and additional examples for highly similar intent pairs.
